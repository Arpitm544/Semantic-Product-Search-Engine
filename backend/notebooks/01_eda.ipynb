```python
import sys
from pathlib import Path
import pandas as pd

# Ensure project root is in sys.path
root_dir = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(root_dir) not in sys.path:
    sys.path.insert(0, str(root_dir))

from src.data.load_data import load_cleaned_products

# Load cleaned products directly from MongoDB (with fallback to local file)
products = load_cleaned_products(prefer_mongo=True)

df = pd.DataFrame(products)
print(f"Total Products Ingested: {len(df)}")
df.head(3)
```

## 1. Catalog Schema & Missing Values Inspection

```python
print("Missing values per column:")
print(df[['id', 'title', 'description', 'category', 'price', 'rating']].isnull().sum())
print(f"\nDuplicate Product IDs: {df['id'].duplicated().sum()}")
```

## 2. Category Distribution

```python
cat_counts = df['category'].value_counts()
print("Top 10 Categories in Catalog:")
print(cat_counts.head(10))
print(f"\nTotal Unique Categories: {len(cat_counts)}")
```

## 3. Price & Rating Distributions

```python
print(df[['price', 'rating']].describe())
```

## 4. Text Representation Length Analysis
Verify that the templated representation text does not exceed the model token context limit (256/512 tokens).

```python
df['title_len'] = df['title'].apply(lambda x: len(x.split()))
df['desc_len'] = df['description'].apply(lambda x: len(x.split()))
df['rep_len'] = df['representation_text'].apply(lambda x: len(x.split()))

print("Word count statistics for representation_text:")
print(df[['title_len', 'desc_len', 'rep_len']].describe())
```
